# Plot Gaussian of scores

```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats
import math
import pickle
```

```python
# read experiment results from the path
result_path = '/Users/yifanlu/PycharmProjects/interpretable_evaluation_metrics/results_dir/celebA_50_USER_dualCS/CelebA5k_furniture15k_vs_CelebA5k_furniture15k_Eyeglasses_Injection_1000/experiment_results_dict.pkl'
with open(result_path, 'rb') as f:
    results = pickle.load(f)
    attr_list = results['attr_list']
    scores_real = results['scores_real']
    scores_gen = results['scores_gen']
    # find index of the target attribute
    target_attr = 'wearing eyeglasses'
    assert target_attr in attr_list
    target_attr_index = attr_list.index(target_attr)
    scores_real_target_attr = scores_real[:, target_attr_index]
    scores_gen_target_attr = scores_gen[:, target_attr_index]

mean_real = np.mean(scores_real_target_attr)
mean_gen = np.mean(scores_gen_target_attr)
variance_real = np.var(scores_real_target_attr)
variance_gen = np.var(scores_gen_target_attr)
sigma_real = math.sqrt(variance_real)
sigma_gen = math.sqrt(variance_gen)
print('DualCS (ours)')
print('attribue: ', target_attr)
print('mean_real: ', mean_real, 'variance_real: ', variance_real)
print('mean_gen: ', mean_gen, 'variance_gen: ', variance_gen)
print('mean_diff: ', mean_real - mean_gen, 'variance_diff: ', variance_real - variance_gen)
```
